```python
class Osoba:
    def __new__(cls, *args, **kwargs):
        return super().__new__(Osoba)
    def __init__(self,imie,nazwisko,wiek,wzrost,waga):
        self.imie = imie
        self.nazwisko = nazwisko
        self.wiek = wiek
        self.wzrost = wzrost
        self.waga = waga
        self.message()
    def __repr__(self):
        return f"osoba: {self.imie} {self.nazwisko}, wiek: {self.wiek} lat"
    def message(self):
        print("to jest obiekt klasy Osoba!")
    def bmi(self):
        return self.waga/(self.wzrost/100)**2
    @staticmethod
    def delta_waga(old,new):
        return new - old
```

```python
os = Osoba("Jan","Kot",45,189,90)
print(os)
print(f"bmi = {os.bmi():.2f}")
print(Osoba.delta_waga(os.waga,101))
```
